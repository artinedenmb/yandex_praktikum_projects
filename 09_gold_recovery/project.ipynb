## Модель

```python
def sMAPE(target, predictions):
    for_one = (abs(target - predictions) / 
    ((abs(target) + abs(predictions)) / 2) * 100)
    result = for_one.sum() / len(target)
    return result

def total_sMAPE(target, predictions):
    rougher_sMAPE = sMAPE(target.iloc[:, 0], predictions[:, 0])
    final_sMAPE = sMAPE(target.iloc[:, 1], predictions[:, 1])
    total_sMAPE = 0.25 * rougher_sMAPE + 0.75 * final_sMAPE
    return total_sMAPE
```

<div class="alert alert-block alert-success">
<b>Успех:</b> Верно.
</div>

Функция для подсчета sMAPE

```python
lst_of_columns.remove('rougher.output.recovery')
lst_of_columns.remove('final.output.recovery')
train_df = train_df.drop(lst_of_columns, axis=1)
full_df = full_df.drop(lst_of_columns, axis=1)
```

Для дальнейшей работы с моделями необходимо удалить из обучающей выборки и полного датасета столбцы, отсутствующие в тестовой выборке.

```python
scaler = StandardScaler()
scaler.fit(train_df.drop(['date'], axis=1))
df_train_scaled = scaler.transform(train_df.drop(['date'], axis=1))
df_train_scaled = pd.DataFrame(df_train_scaled)
print(train_df.columns[1:])
df_train_scaled.columns = train_df.columns[1:]
df_train_scaled.info()
```

Проведем масштабирование признаков для более точной работы модели

```python
sMAPE_scorer = make_scorer(total_sMAPE, greater_is_better=False)
```

<div class="alert alert-block alert-success">
<b>Успех:</b> Скорер определен корректно.
</div>

Создадим оценку, на которую будем ориентироваться при выборе модели

```python
train_X = df_train_scaled.drop(
        ['rougher.output.recovery', 'final.output.recovery'], axis=1)
train_Y = train_df[['rougher.output.recovery', 'final.output.recovery']]
```

```python
model = LinearRegression()
scores = cross_val_score(model, X=train_X, y=train_Y,
                         scoring=sMAPE_scorer, cv=5)
final_score = -scores.mean()
print('sMAPE:', final_score)
```

```python
model_2 = RandomForestRegressor(random_state=12345)
parametrs = {'n_estimators': range(40, 41, 1),
             'max_depth': range(6, 7, 1)}

grid = GridSearchCV(model_2, parametrs, cv=5, scoring=sMAPE_scorer)
grid.fit(train_X, train_Y)
print('Параметры:')
print(grid.best_params_)
print('sMAPE:')
print(grid.best_score_ * (-1))
```

<div class="alert alert-block alert-success">
<b>Успех:</b> Отлично, что для подбора парамтеров использовался метод оснвоанный на кросс-валидации!
</div>

Модель случайного леса оказалась более точной

```python
model_3 = grid.best_estimator_
```

```python
test_f = test_df.drop('date', axis=1)
test_t = test_df.merge(full_df, on='date', how='left')[['rougher.output.recovery', 'final.output.recovery']]
```

```python
predictions = model_3.predict(test_f)
result = total_sMAPE(test_t, predictions)
print('sMAPE:', result)
```
