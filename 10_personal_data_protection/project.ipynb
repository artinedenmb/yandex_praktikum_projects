## Умножение матриц

В этом задании вы можете записывать формулы в *Jupyter Notebook.*

Чтобы записать формулу внутри текста, окружите её символами доллара \\$; если снаружи —  двойными символами \\$\\$. Эти формулы записываются на языке вёрстки *LaTeX.* 

Для примера мы записали формулы линейной регрессии. Можете их скопировать и отредактировать, чтобы решить задачу.

Работать в *LaTeX* необязательно.

Обозначения:

- $X$ — матрица признаков (нулевой столбец состоит из единиц)

- $y$ — вектор целевого признака

- $P$ — матрица, на которую умножаются признаки

- $w$ — вектор весов линейной регрессии (нулевой элемент равен сдвигу)

Предсказания:

$$
a = Xw
$$

Задача обучения:

$$
w = \arg\min_w MSE(Xw, y)
$$

Формула обучения:

$$
w = (X^T X)^{-1} X^T y
$$

**Ответ:** не изменится

**Обоснование:** 

Для того, чтобы доказать, что качество линейной регрессии не изменится, необходимо доказать, что выражение $ a_1 = XPw = XP @((XP).T @ (XP))^{-1} @ (XP).T @ y $ равно $ a = Xw = X @(X.T @ X)^{-1} @ X.T @ y $,
для этого применим свойство матриц $ (AB).T = B.T @ A.T $

 $ XP @ ((XP).T @ XP)^{-1} @ (XP).T @  y = XP @ (P.T @  X.T @ XP)^{-1} @  P^T @  X^T @ y = XP @ (P.T @ (X.T @ XP))^{-1} @ P.T  @ X.T @ y $
 
 Теперь можно использовать следующее свойство матриц: 
 
 $ (AB)^{-1} = B^{-1} @ A^{-1} $, при том, что обратной матрицы $ X^{-1} $ не существует
 
$ XP @ (P.T @ (X.T @ XP))^{-1} @ P.T @X.T @ y = XP @ (X.T @ XP)^{-1} @ (P.T)^{-1} @ P.T @ X.T @ y  = XP @ P^{-1} @ (X.T  @ X)^{-1} @ (P.T)^{-1} @ P.T @ X.T @ y $
 
 Заменим $ P @ P^{-1} на E, так как A @ A^{-1} = E $
 
 $ XP @ P^{-1} @ (X.T @ X)^{-1} @ (P.T)^{-1} @ P.T @ X.T @ y = X @ E @ (X.T @ X)^{-1} @ E @ X.T @ y $
 
 Очевидно, что AE = EA = A, следовательно
 
$ X @ E @ (X.T @ X)^{-1} @ E @ X.T @ y = X @ (X.T @ X)^{-1} @ X.T $
 
 Тогда выражение a_1 действительно равно a.

## Проверка алгоритма

Сначала посмотрим на качество модели при неизмененных признаках.

```python
model = LinearRegression()
model.fit(features, target)
predictions = model.predict(features)
r2 = r2_score(target, predictions)
print(r2)
```

Теперь сгененируем рандомную матрицу

```python
np.random.seed(42)
random_matrix = np.random.randint(1, 40, size=(4, 4))
print(random_matrix)
```

Убедимся, что сгенерированная матрица обратимая

```python
while np.linalg.det(random_matrix) == 0:
    random_matrix = np.random.randint(1, 40, size=(4,4))
```

Произведем преобразоание данных

```python
features_modified = features @ random_matrix
```

```python
print(features_modified.shape)
```

```python
features_modified.head()
```

Из преобразованных данных нельзя догататься о конфиденциальной информации клиентов страховой компании.

Теперь посчитаем r2 для преобразованных данных.

```python
model = LinearRegression()
model.fit(features_modified, target)
predictions_modified = model.predict(features_modified)
r2_modified = r2_score(target, predictions_modified)
print(r2_modified)
```

Показатель r2 совершенно не изменился. Он примерно равен 0.42 как до преобразования, так и после. Качество модели не ухудшилось, а данные стали защищенными.
