В столбце locality_name встречается большое количество дубликатов, их можно удалить следующим образом.

```python
data['locality_name'] = data['locality_name'].fillna('Не указано')
list_of_not_names = ['посёлок городского типа ', 'поселок городского типа ', 'коттеджный посёлок ', 'городской посёлок ', 'поселок ']
for i in range(len(data.loc[:, 'locality_name'])):
    for not_name in list_of_not_names:
        if not_name in data.loc[i, 'locality_name']:
            the_name = data.loc[i, 'locality_name'].split(not_name)
            data.loc[i, 'locality_name'] = 'посёлок ' + the_name[-1]
            break
print(data['locality_name'].unique())
```

```python
for locality_name in data['locality_name'].unique():
    median = data.loc[data['locality_name'] == locality_name, 'ceiling_height'].median()
    data.loc[(data['ceiling_height'].isna()) & (data['locality_name'] == locality_name), 'ceiling_height'] = median
    
data.loc[data['ceiling_height'].isna(), 'ceiling_height'] = data['ceiling_height'].median()
```

Пропуски в столбце ceiling_height можно заполнить медианным значением в зависимости от населенного пункта. Оставшиеся значения можно просто заменить медианными по Ленинградской области.

```python
data = data.dropna(subset=['living_area'])
```

Выведем таблицу корреляции.

```python
for i in ['total_area', 'living_area', 'kitchen_area']:
    data[i + '_category'] = np.NaN
    D_indexer = data[i].describe()['25%']
    C_indexer = data[i].describe()['50%']
    B_indexer = data[i].describe()['75%']
    data.loc[data[i] < D_indexer, i + '_category'] = 'D'
    data.loc[(data[i] < C_indexer) & (data[i] >= D_indexer), i + '_category'] = 'C'
    data.loc[(data[i] < B_indexer) & (data[i] >= C_indexer), i + '_category'] = 'B'
    data.loc[data[i] >= B_indexer, i + '_category'] = 'A'
```

Создадим столбцы с категориями для параметров с большим разбросом значений, далее это будет использоваться для создания сводных таблиц.

```python
def corr_param(param):
    return f'for {param} it is {data["last_price"].corr(data[param])}'

def graphics(data, param): 
    graphics1 = data.plot(kind='scatter', x='last_price', y=param, alpha=0.2, title=param + '_scatterplot') 
    graphics2 = data.plot(kind='hexbin', x='last_price', y=param, gridsize=30, figsize=(8, 6),
                          sharex=False, grid=True, title=param + '_hexbinplot')

def pivot_tables(param):
    pivot_data = data.pivot_table(index=param, values='last_price', aggfunc='median')
    pivot_data.columns = ['median']
    return pivot_data

def new_plot(param):
    to_plot = data.groupby(param).median()['last_price'].reset_index().sort_values(by='last_price')
    to_plot.plot(kind='bar', x=param, y='last_price', figsize=(9, 12), title=param)
    if param == 'rooms':
        to_plot = data[data['rooms'] < 7].groupby(param).median()['last_price'].reset_index().sort_values(by='last_price')
        to_plot.plot(kind='bar', x=param, y='last_price', figsize=(9, 12), title=param+' normal values')
```

<div class="alert alert-block alert-success">✔️
    

__Комментарий от ревьюера №1__

Отличная функция, ты молодец =)
</div>

```python
for i in ['total_area', 'living_area', 'kitchen_area', 'rooms', 'city_centers_nearest_km']:
    print(corr_param(i))
```

Высчитаем корреляцию для количественных параметров. Видно, что из них фактором, наиболее сильно влияющим на цену, является общая площадь. Фактором, оказывающим наименьшее влияние, является удаленность от центра. (определяем по модулю)

```python
for i in ['total_area_category', 'living_area_category', 'kitchen_area_category', 'rooms', 'type_of_floor', 'week_day_exposition',
          'month_of_exposition', 'year_of_exposition']:
    print(pivot_tables(i))
    print()
    new_plot(i)
```

Цена закономерно увеличивается для категорий общей, жилой площади и площади кухни. Значения в категории A очень большие, так как там разброс значений очень большой.
